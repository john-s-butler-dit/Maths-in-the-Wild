## Predicting Many Years in the Future



```python
# Set the timestep for the simulation (1 year per step)
timestep = 1

# Create an array of years from 1895 to 1935 (inclusive) with the given timestep
time = np.arange(1895, 1936, timestep)

# Calculate the number of time steps (years)
N = len(time)

# Initialize an array to store the hare population at each time step
H = np.zeros(N)
```

### Coding Exercise 2.3.1: Recurrent Equation with a __for__ Loop

The hare population can be modeled using a __recurrent equation__, which references its previous value:
$$H= H + (r_{Birth}×H). $$

This can be code using an array and a __for__ loop:

$$H[i+1]= H[i] + (r_{birth} * H[i]), $$
where $i$ starts at $0$ and goes to the length of the the array minus 1.

### Coding Exercise 2.3.1: Recurrent Equation with a __for__ Loop

Given:

- Initial hare population in 1895: $H=85$
- Birth rate: $r_{\text{Birth}} = 0.2$

Write code to implement the recurrent equation to simulate the hare population using a for loop.


```python
# Set the initial hare population (in tens of thousands)
H[0] = 85

# Set the birth rate for hares (growth rate per year)
r_Birth = 0.2

# Loop through each time step to compute the hare population using a recurrence relation
for i in range(0, N - 1):
    
    ### Solution 3.3: Recurrent Equation
    # Update the hare population for the next year based on the current population
    # This uses exponential growth: H[i+1] = H[i] + (growth rate × current population)
    H[i + 1] = H[i] + (r_Birth * H[i])
    ###
    
```

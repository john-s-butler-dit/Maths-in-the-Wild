## Modelling the  Lynx Population

To model the lynx population we use a __recurrent equation__ where we use the current population to predict the future population again and again and again.


```python
# Set the timestep for the simulation (1 year per step)
timestep = 1

# Create an array of years from 1895 to 1935 (inclusive) with the given timestep
time = np.arange(1895, 1936, timestep)

# Calculate the number of time steps (years)
N = len(time)

# Initialize an array to store the lynx population at each time step
L = np.zeros(N)

```

## Recurrent Equation 

The lynx population can be modeled using a __recurrent equation__, which references its previous value:
$$L= L - (r_{Death}×L). $$


This can be coded as a for loop using an array: 
$$L[i+1]= L[i] - (r_{Death} * L[i]), $$
where $i$ starts at $0$ and goes to the length of the array minus 1.

### Coding Exercise 3.3: Recurrent Equation using a __for__ loop
Given:

- Initial lynx population in 1895: $L=51$
- Death rate: $r_{\text{Death}} = 0.1$

Write code to implement the recurrent equation to simulate the lynx population using a for loop.

```python
L[0]=51 # Start Population

# Set the birth rate for hares (growth rate per year)
r_Death = 0.1

# Loop through each time step to compute the hare population using a recurrence relation
for i in range(0, N - 1):
    
    ### Solution 4.3: Recurrent Equation
    # Update the lynx population for the next year based on the current population
    L[i + 1] = L[i] - (r_Death * L[i])
    ###
 
```

```python
print("Lynx Population in 1935:")
print(round(L[-1]))
```

## Change of Death Rate
The output below states the lynx population in 1935 for a Death rate to 0.05.
The plot shows the simulated lynx population from 1895 tp 1935 for a Death rate to 0.05.

### Reflective Exercise 3.5: Change of Death Rate
Compare the lynx population in 1935 (with a death rate of 0.05) to the population trend shown in the plot when the birth rate is 0.1.

What are the similarities and differences between the population value in 1935 and the overall population trend in the plot for the birth rate of 0.1?

```python
# Set the lynx death rate per year
r_death = 0.05

# Loop through each time step from 0 to N-2 (since we use i+1)
for i in range(0, N - 1):
    # Update the lynx population by subtracting deaths
    # Deaths are calculated as r_death * current population
    L[i + 1] = L[i] - (r_death * L[i])


# Print the lynx population at the final year (1935)
print("Lynx Population in 1935:")
print(round(L[-1]))  # Round to nearest whole number
```

```python
# Create a new figure with a specified size (8 inches wide, 4 inches tall)
fig = plt.figure(figsize=(8, 4))

# Plot the Lynx population data (L) against time
# 's:' means square markers with a dotted line
# color='red' sets the line color to red
# label='Lynx' adds a label for the legend
plt.plot(time, L, 's:', color='red', label='Lynx')

# Set the x-axis label to indicate time in years
plt.xlabel('Time (yrs)')

# Set the y-axis label to indicate population in tens of thousands
plt.ylabel('Population (tens of thousands)')

# Display the plot
plt.show()
```
